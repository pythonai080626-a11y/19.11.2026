```python
shifts = pd.read_csv('shifts.csv', index_col = 'Day')
shifts['Total'] = shifts['Omer'] + shifts['Noa'] + shifts['Maya'] + shifts['Yael']
shifts['Total'] = shifts[['Noa', 'Omer', 'Maya', 'Yael']].sum(axis=1)
shifts['Total'] = shifts[shifts.columns[:-1]].sum(axis=1)
shifts['Total'] = shifts[[col for col in shifts.columns if col != 'Total']].sum(axis=1)
display(shifts)
```

```python
shifts = pd.read_csv('shifts.csv', index_col = 'Day')
shifts['Total'] = shifts[['Noa', 'Omer', 'Maya', 'Yael']].sum(axis=1)
display(shifts)
shifts.drop(axis=1, index='Sunday')
shifts.drop(axis=1, index=['Tuesday', 'Monday'])
```
